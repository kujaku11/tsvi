# Getting Started With TSVI

This notebook will show you how to get the mth5 time series viewer running.

## Get the code

The code for tsvi is on GitHub and can be readily cloned via Git. The following command should work straight from the command prompt in Linux or MacOS. For Windows, you can open [Git Bash](https://gitforwindows.org/) first and run the line there.

```python
git clone https://github.com/kujaku11/tsvi.git
```

Next, create an environment to work from using the following command. On Windows, you can use [Anaconda Prompt](https://docs.anaconda.com/anaconda/install/windows/) or [Miniforge](https://github.com/conda-forge/miniforge) as your console.

```python
conda create -n tsvi python=3.9
```

Then activate the environment with the following command:

```python
conda activate tsvi
```

Install the dependencies and jupyter-lab as well:

```python
cd tsvi/
pip install -e .
pip install jupyter-lab
```

Allow jupyter to see your environment:

```python
python -m ipykernel install --user --name=tsvi
```

At this point, you should be able to start executing .ipynb notebooks from jupyter-lab. This is as simple as running:

```python
jupyter notebook
```

Other useful installs for this project include:

```python
pip install git+https://github.com/simpeg/aurora.git
pip install git+https://github.com/kujaku11/mth5.git
```

MTH5 hosts a variety of tools useful for processing raw MT data files of many formats into a single self-descriptive format that can be readily processed by tsvi and more specifically the mth5_viewers. Aurora helps estimate single station and remote reference transfer functions from the data provided by MTH5 as well.
